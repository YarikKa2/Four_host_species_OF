## Flat view

```python

INPUT_PATH = Path(os.path.dirname(os.getcwd()).replace('\\', '/') + 
                  '/full_view/Hamster_fullview.png')   # change this
OUTPUT_PATH = Path(os.path.dirname(os.getcwd()).replace('\\', '/') + 
                   '/full_view/Hamster_fullview.png') # TIFF or PNG recommended

SIGMA = 1500        # Try 500, 700, 900, 1200, or 1500
STRENGTH = 0.75    # 0 = unchanged; 1 = strongest correction
```

```python
try:
    # PIL opens files using standard Python file handlers (safest for paths)
    with Image.open(INPUT_PATH) as pil_img:
        # Convert to RGB (in case it is CMYK, Grayscale, or WebP)
        rgb_img = pil_img.convert("RGB")
        
        # Convert the PIL image into a NumPy array
        img = np.array(rgb_img)
        
        # Convert RGB to OpenCV's native BGR format
        img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
        
    print(f"✅ Success! PIL successfully parsed and converted the image. Shape: {img.shape}")
except Exception as e:
    print(f"❌ Completely unreadable file. Error: {e}")
```

```python
has_alpha = img.ndim == 3 and img.shape[2] == 4
bgr = img[:, :, :3] if img.ndim == 3 else cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
alpha = img[:, :, 3] if has_alpha else None

img.shape
```

```python
original_dtype = bgr.dtype
max_value = np.iinfo(original_dtype).max if np.issubdtype(original_dtype, np.integer) else 1.0

# Convert to float so calculations do not clip or overflow.
bgr_float = bgr.astype(np.float32) / max_value

# Correct only brightness/lightness: helps preserve H&E color balance.
lab = cv2.cvtColor(bgr_float, cv2.COLOR_BGR2LAB)
lightness = lab[:, :, 0]

# Smooth illumination field: use kernel (0, 0) and control blur by sigma.
illumination = cv2.GaussianBlur(
    lightness,
    ksize=(0, 0),
    sigmaX=SIGMA,
    sigmaY=SIGMA,
    borderType=cv2.BORDER_REFLECT
)

# Estimate reference brightness from tissue-like pixels, excluding white background.
tissue_mask = lightness < 95.0
reference = np.median(illumination[tissue_mask]) if tissue_mask.any() else np.median(illumination)

# Multiplicative flat-field correction with conservative gain limits.
gain = reference / np.maximum(illumination, 1e-6)
gain = np.power(gain, STRENGTH)
gain = np.clip(gain, 0.65, 1.55)

corrected_lightness = np.clip(lightness * gain, 0, 100)
lab[:, :, 0] = corrected_lightness

corrected_float = cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)
corrected_bgr = np.clip(corrected_float * max_value, 0, max_value).astype(original_dtype)

# Restore alpha, if the source has transparency.
corrected = np.dstack([corrected_bgr, alpha]) if has_alpha else corrected_bgr
```

```python
def show_bgr(img, title, max_width=1200):
    display_img = img[:, :, :3] if img.ndim == 3 else img

    height, width = display_img.shape[:2]
    scale = min(1.0, max_width / width)

    if scale < 1:
        display_img = cv2.resize(
            display_img,
            (int(width * scale), int(height * scale)),
            interpolation=cv2.INTER_AREA
        )

    plt.figure(figsize=(16, 10))
    plt.imshow(cv2.cvtColor(display_img, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.axis("off")
    plt.show()

show_bgr(bgr, "Original")
show_bgr(corrected, f"Flat-field corrected — sigma={SIGMA}, strength={STRENGTH}")
```

```python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

success = cv2.imwrite(str(OUTPUT_PATH), corrected)

if not success:
    raise RuntimeError(f"Could not save output: {OUTPUT_PATH}")

print(f"Saved corrected image: {OUTPUT_PATH.resolve()}")
print(f"Output size: {OUTPUT_PATH.stat().st_size / 1024 / 1024:.2f} MB")
```
